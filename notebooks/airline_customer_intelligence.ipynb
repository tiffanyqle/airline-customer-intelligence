```python
#create clustering dataset
cluster_data = df_clean[rating_columns].dropna().copy()

cluster_data.shape
```

```python
#average and spread of each rating
cluster_data.describe().T
```

```python
%pip install scikit-learn
from sklearn.preprocessing import StandardScaler
```

```python
#standardizing the variables
scaler = StandardScaler()

cluster_scaled = scaler.fit_transform(cluster_data)

cluster_scaled.shape
```

```python
#using elbow method to test several values of k (number of clusters)
#and measures how tightly passengers fit within each cluster
from sklearn.cluster import KMeans

inertia = []

k_values = range(2, 9)

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(cluster_scaled)
    inertia.append(kmeans.inertia_)

list(zip(k_values, inertia))
```

```python
#graphs it
%pip install matplotlib
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(k_values, inertia, marker="o")

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means Clustering")

plt.show()
```

```python
#building the 4 cluster model
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(cluster_scaled)
```

```python
#sees how many passengers end up in each cluster
pd.Series(cluster_labels).value_counts().sort_index()
```

```python
#attach each passenger's cluster number back to the clustering dataset
cluster_profiles = cluster_data.copy()

cluster_profiles["Cluster"] = cluster_labels
```

```python
#calculates the average rating for every service within each cluster
cluster_means = cluster_profiles.groupby("Cluster").mean()

cluster_means.T
```

```python
#create prediction target
#turns satisfied into 1 and neutral/dissatisfied into 0
df_model = df_clean.copy()

df_model["Satisfied"] = (
    df_model["Satisfaction"] == "Satisfied"
).astype(int)

df_model[["Satisfaction", "Satisfied"]].head(10)
```

```python
#creating the first predictive model
model_features = [
    "Gender",
    "Age",
    "Customer Type",
    "Type of Travel",
    "Class",
    "Flight Distance",
    "Departure Delay",
    "Arrival Delay"
] + rating_columns

X = df_model[model_features].copy()
y = df_model["Satisfied"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#seperating numeric and categorical variables
numeric_features = [
    "Age",
    "Flight Distance",
    "Departure Delay",
    "Arrival Delay"
] + rating_columns

categorical_features = [
    "Gender",
    "Customer Type",
    "Type of Travel",
    "Class"
]

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
#preprocessor
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            drop="first"
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
#splitting the data, using 80% to train the model
#and 20% holds back to test it
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training passengers:", len(X_train))
print("Testing passengers:", len(X_test))
```

```python
#training a logistic regression model
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000))
    ]
)

logistic_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = logistic_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")
```

```python
#classification report
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
#pulling out all the variable names after preprocessing
feature_names = logistic_model.named_steps[
    "preprocessor"
].get_feature_names_out()

coefficients = logistic_model.named_steps[
    "classifier"
].coef_[0]

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

feature_importance["Absolute Coefficient"] = (
    feature_importance["Coefficient"].abs()
)

feature_importance = feature_importance.sort_values(
    "Absolute Coefficient",
    ascending=False
)

feature_importance.head(15)
```

```python
#calculates ROC-AUC
#measures how well the model distinguishes satisfied passengers
#from neutral/dissatisfied across different classification thresholds
from sklearn.metrics import roc_auc_score

y_prob = logistic_model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", round(roc_auc, 3))
```

```python
#training random forest comparison model
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

random_forest_model.fit(X_train, y_train)
```

```python
rf_pred = random_forest_model.predict(X_test)
rf_prob = random_forest_model.predict_proba(X_test)[:, 1]

rf_accuracy = accuracy_score(y_test, rf_pred)
rf_auc = roc_auc_score(y_test, rf_prob)

print("Random Forest Accuracy:", round(rf_accuracy * 100, 2), "%")
print("Random Forest ROC-AUC:", round(rf_auc, 3))
```

## Business Recommendations

### 1. Strengthen the Digital Passenger Journey
Online boarding and in-flight Wi-Fi consistently emerged as prominent indicators of passenger satisfaction. The airline should prioritize a streamlined digital boarding experience, reliable onboard connectivity, and simplified online booking, with targeted support for passenger segments experiencing digital friction.

### 2. Personalize Interventions for Economy Passenger Experience Gaps
Customer segmentation revealed that lower-satisfaction Economy passengers experience different types of friction. Some passengers reported weaker comfort and amenity experiences, while others experienced greater service and operational friction. The airline should tailor customer experience improvements to these distinct needs rather than applying the same intervention to all dissatisfied passengers.

### 3. Develop a First-Time Passenger Experience Strategy
First-time passengers had substantially lower satisfaction than returning passengers and were more concentrated in the lowest-satisfaction customer segment. The airline should consider targeted pre-trip guidance, digital onboarding, and post-trip engagement designed to reduce friction for new customers.

### 4. Use Delay Severity to Trigger Proactive Service Recovery
Passenger satisfaction declined as departure and arrival delays became more severe. The airline should use delay severity as a trigger for proactive communication and customer support, such as updated connection information, rebooking assistance, and appropriate service-recovery interventions.
